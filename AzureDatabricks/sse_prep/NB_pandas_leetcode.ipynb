```python
# Is val equal to element in list? If so, remove in place (replace) and return count of not equal
nums = [1,2,3,3,4]

def remove_element(nums, val):
    k = 0
    for i in range(len(nums)):
        if nums[i] != val:
            nums[k] = nums[i]
            k += 1
    return k
print(remove_element(nums, 3))
```

```python
from typing import List

# Remove duplicates

def removeDuplicates(nums: List[int]) -> int:
    # remove in place -> replace element
    # distinct elements in same order as original
    # return count of distinct elements
    # [1,2,3] -> is 1 in [2, 3]

    k = 0

    for i in range(len(nums)):
        if nums[i] not in nums[i+1:]:
            nums[k] = nums[i]
            k += 1
    return k
nums = [0,0,1,1,1,2,2,3,3,4]
print(removeDuplicates(nums))
```

```python
# Return the element that appears more than n/2 times in array
from typing import List

def majorityElement(nums: List[int]) -> int:
        # Boyer–Moore Voting Algorithm
        # 3/2 = 1,33 -> ex. 2 times
        # 7/2 = >3 -> ex. 4 times
        count = 0
        candidate = None

        for num in nums:
            if count == 0:
                candidate = num
            count += (1 if num == candidate else -1)

        return candidate  
nums = [2,2,1,1,1,2,2]
print(majorityElement(nums))      
```

```python
# Return the element that appears more than n/2 times in array
def majorityElement(nums):
    nums.sort()
    return nums[len(nums) // 2]

nums = [3, 2, 3]
print(majorityElement(nums))
```

```python
# Buy and sell stock and return max profit
from typing import List
# same as sell stock I but save stock
# after profit making set min_price and max_prift to default.
def maxProfit(prices: List[int]) -> int:
    min_price = float('inf') # 1, None, 3
    take_profit = 0
    total_profit = 0 # 4, 3

    for price in prices:
        if price < min_price:
            min_price = price
        elif price - min_price > take_profit:
            profit = price - min_price
            total_profit += profit
            min_price = price
    return total_profit

print(maxProfit([1,2,3,4,5]))
```

```python
# Convert roman numeric values to int
def romanToInt(s: str) -> int:
    # break down input "I I I" or "I V"
    # search input char from map
    roman = {'I':1, 'V':5, 'X':10, 'L':50,
                'C':100, 'D':500, 'M':1000}
    
    result = 0
    for i in range(len(s)):
        curr = roman[s[i]] # search input char one by one
        next_val = roman[s[i+1]] if i+1 < len(s) else 0
        
        if curr < next_val:
            result -= curr
        else:
            result += curr
    
    return result
```
