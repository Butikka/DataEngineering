```python
# Is val equal to element in list? If so, remove in place (replace) and return count of not equal
nums = [1,2,3,3,4]

def remove_element(nums, val):
    k = 0
    for i in range(len(nums)):
        if nums[i] != val:
            nums[k] = nums[i]
            k += 1
    return k
print(remove_element(nums, 3))
```

```python
from typing import List

# Remove duplicates

def removeDuplicates(nums: List[int]) -> int:
    # remove in place -> replace element
    # distinct elements in same order as original
    # return count of distinct elements
    # [1,2,3] -> is 1 in [2, 3]

    k = 0

    for i in range(len(nums)):
        if nums[i] not in nums[i+1:]:
            nums[k] = nums[i]
            k += 1
    return k
nums = [0,0,1,1,1,2,2,3,3,4]
print(removeDuplicates(nums))
```

```python
# Return the element that appears more than n/2 times in array
from typing import List

def majorityElement(nums: List[int]) -> int:
        # Boyer–Moore Voting Algorithm
        # 3/2 = 1,33 -> ex. 2 times
        # 7/2 = >3 -> ex. 4 times
        count = 0
        candidate = None

        for num in nums:
            if count == 0:
                candidate = num
            count += (1 if num == candidate else -1)

        return candidate  
nums = [2,2,1,1,1,2,2]
print(majorityElement(nums))      
```

```python
# Return the element that appears more than n/2 times in array
def majorityElement(nums):
    nums.sort()
    return nums[len(nums) // 2]

nums = [3, 2, 3]
print(majorityElement(nums))
```

```python
# Buy and sell stock and return max profit
from typing import List
# same as sell stock I but save stock
# after profit making set min_price and max_prift to default.
def maxProfit(prices: List[int]) -> int:
    min_price = float('inf') # 1, None, 3
    take_profit = 0
    total_profit = 0 # 4, 3

    for price in prices:
        if price < min_price:
            min_price = price
        elif price - min_price > take_profit:
            profit = price - min_price
            total_profit += profit
            min_price = price
    return total_profit

print(maxProfit([1,2,3,4,5]))
```

```python
# Convert roman numeric values to int
def romanToInt(s: str) -> int:
    # break down input "I I I" or "I V"
    # search input char from map
    roman = {'I':1, 'V':5, 'X':10, 'L':50,
                'C':100, 'D':500, 'M':1000}
    
    result = 0
    for i in range(len(s)):
        curr = roman[s[i]] # search input char one by one
        next_val = roman[s[i+1]] if i+1 < len(s) else 0
        
        if curr < next_val:
            result -= curr
        else:
            result += curr
    
    return result
```

```python
# Comibne 2 numbers in order to have same value as target
def twoSum(self, nums: List[int], target: int) -> List[int]:
    # [0,1] -> output list of indeces
    # [2,7,11] -> 2,7 and 2,11 and 7,11 

    pair_idx = {}

    for i, num in enumerate(nums):
        if target - num in pair_idx:
            return [i, pair_idx[target - num]]
        pair_idx[num] = i # num adds 2 as key to dict and i as index {2: 0}
```

```python
from typing import List
# Write a function to find duplicate records based on email.
records = [
    {"id": 1, "name": "Alice", "email": "alice@example.com"},
    {"id": 2, "name": "Bob", "email": "bob@example.com"},
    {"id": 3, "name": "Alice", "email": "alice@example.com"}
]

# records["email"] -> 
# equality
# hashset() -> save email as key
# i in hashset() -> if true then duplicate
# add dubples to a list

def find_duplicates(records) -> List["str"]:
    
    hashset = set() # dict of keys
    dupl_list = [] # final list of dublicate emails

    for i in records:
        if i["email"] not in hashset:
            hashset.add(i["email"])
        else:
            dupl_list.append(i["email"])
    return dupl_list

print(find_duplicates(records))
```

```python
# Write Python code that:

# Processes each record in real time.

# Maintains a rolling average of the last 2 values.

stream = [
    {"event_time": "2025-08-12 10:00", "value": 10},
    {"event_time": "2025-08-12 10:01", "value": 20},
    {"event_time": "2025-08-12 10:02", "value": 15}
]

# While loop
# 15, 20 -> 15+20 / 2 = 17,5
# i["value"]
# pop() -> twice then 2 last ones

def rolling_avg(stream):
    rolling_averages_while = []
    i = 0
    while i < len(stream):
        if i >= 1:
            average = (stream[i]['value'] + stream[i-1]['value']) / 2
            rolling_averages_while.append(average)
        i += 1
    return rolling_averages_while

print(rolling_avg(stream))
```

```python
from pyspark.sql.functions import col
# Perform an inner join in Python (without Pandas) on users.id = transactions.user_id and return:
# return this: [{"name": "Alice", "amount": 100}, ...]
users = [
    {"id": 1, "name": "Alice"},
    {"id": 2, "name": "Bob"}
]

transactions = [
    {"user_id": 1, "amount": 100},
    {"user_id": 1, "amount": 200},
    {"user_id": 2, "amount": 50}
]

# df.join(table1, table1.col == table2.col, "inner") -> spark
# df 

def inner_join(users, transactions):
    df_users = spark.createDataFrame(users)
    df_transactions = spark.createDataFrame(transactions)
    df_joined = df_users.join(df_transactions, col("id")== col("user_id"), "inner")
    df_final = df_joined.select("name", "amount")
    return df_final

df = inner_join(users, transactions)
df.show()
```

```python
from datetime import datetime
# Write Python code to:

# Keep only the latest record per id based on timestamp.
data = [
    {"id": 1, "timestamp": "2025-08-12T09:00:00", "value": 10},
    {"id": 1, "timestamp": "2025-08-12T10:00:00", "value": 20},
    {"id": 2, "timestamp": "2025-08-12T09:30:00", "value": 15}
]

latest_records = {}
for record in data:
    record_id = record['id']
    record_timestamp = datetime.fromisoformat(record['timestamp'])
    
    # Check if the ID is new or the current record is newer
    if record_id not in latest_records or record_timestamp > datetime.fromisoformat(latest_records[record_id]['timestamp']): # {1: {...}}
        latest_records[record_id] = record

# Convert the dictionary values back to a list
result = list(latest_records.values())
print(result)
```

```python
# Task: Merge dict1 and dict2 so that common keys have summed values, and unique keys are kept as is.
dict1 = {"apple": 10, "banana": 5, "orange": 8}
dict2 = {"apple": 3, "banana": 7, "pear": 4}

def merge_dicts(dict1, dict2):
    merged_dict = dict1.copy()

    for key, value in dict2.items():
        merged_dict[key] = merged_dict.get(key, 0) + value
    return merged_dict

print(merge_dicts(dict1, dict2))
```
